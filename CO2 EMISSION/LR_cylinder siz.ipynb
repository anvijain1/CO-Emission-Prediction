```python
import pandas as pd 
import matplotlib.pyplot as plt
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
import numpy as np
```

```python
url= "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-ML0101EN-SkillsNetwork/labs/Module%202/data/FuelConsumptionCo2.csv"
```

```python
df = pd.read_csv(url)
```

```python
engine_size = df['ENGINESIZE']
co2_emmision = df['CO2EMISSIONS']
```

```python
plt.scatter(engine_size , co2_emmision )
plt.xlabel('Engine Size')
plt.xlabel('CO2 Emission')
plt.show()
```

```python
df.info()
```

```python
model = LinearRegression()
```

```python
inputs = engine_size
targets = co2_emmision
```

```python
inputs = df[['ENGINESIZE']]
targets = df['CO2EMISSIONS']

model.fit(inputs, targets)
```

```python
X = [[2] , [3], [4]]
```

```python
predictions = model.predict(X)
print(predictions)
```

```python
predictions = model.predict(df[['ENGINESIZE']])

rmse = np.sqrt(mean_squared_error(df['CO2EMISSIONS'], predictions))

print(rmse)
```

```python
plt.scatter( engine_size , co2_emmision , alpha = 0.5)
plt.plot(engine_size, predictions, 'r' )
plt.xlabel('engine size')
plt.ylabel('CO2 Emmision')
plt.grid(color = 'gray' , linestyle = ':' , linewidth = 1)
```
