```python
colors = {'D':'red','Z':'blue','X':'green','E':'yellow'}

for fuel, color in colors.items():
    subset = df[df['FUELTYPE'] == fuel]
    plt.scatter(subset['ENGINESIZE'], subset['CO2EMISSIONS'], color=color, alpha = 0.3)

plt.xlabel("Engine Size")
plt.ylabel("CO2 Emissions")
plt.show()
```
